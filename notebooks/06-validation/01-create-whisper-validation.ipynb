```python
from ibis import _
from ibis import selectors as s
import pandas as pd

import src
from src.load import DataLoader
```

```python
pd.set_option("display.max_colwidth", None)
```

```python
dl = DataLoader()
```

```python
videos = (
    dl.videos(filtered=True)
    .filter(_.video_duration.between(65, 85))
    .join(dl.channels(), "channel_id")
    .select(s.cols("party", "video_id"))
    .to_pandas()
)

videos = videos.groupby(["party"]).sample(n=1, random_state=42)
```
