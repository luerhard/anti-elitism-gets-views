```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.orm import Query

import src
from src.data.models import Channel
from src.data.models import Comment
from src.data.models import Video
```

```python
engine = create_engine(src.PS_ENGINE)
```

# Channels

```python
query = Query(Channel)

with engine.connect() as con:
   df = pd.read_sql(query.statement, con) 
```

```python
df.to_parquet(src.PATH / "data/raw/yt_metadata/channels.parquet.gzip", compression="gzip")
```

# Comments

```python
query = Query(Comment)

with engine.connect() as con:
   df = pd.read_sql(query.statement, con) 
```

```python
df.datetime_upload = pd.to_datetime(df.datetime_upload)
```

```python
df = df.drop("is_valid", axis=1)
```

```python
df.to_parquet(src.PATH / "data/raw/yt_metadata/comments.parquet.gzip", compression="gzip")
```
