```python
import warnings

import librosa
from silero_vad import get_speech_timestamps
from silero_vad import load_silero_vad

import src

warnings.filterwarnings(
    "ignore",
    category=UserWarning,
    message="PySoundFile failed*",
)
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
folder = src.PATH / "data/raw/yt/@FDP/videos"

model = load_silero_vad()

for file in folder.iterdir():
    audio, sr = librosa.load(file, sr=16_000)
    speech_timestamps = get_speech_timestamps(
        audio,
        model,
        sampling_rate=16_000,
        return_seconds=True,
    )
    _break = 0
    for ts in speech_timestamps:
        if (ts["end"] - ts["start"]) > 30:
            _break += 1

    if _break > 0:
        print(_break)
        print(file)
        print("-" * 30)
```

```python
str1 = "Immer wieder gibt es da Zuwächse. Und zum anderen, was nicht so erfreulich ist, aber auch Fakt der Bürokratie und den langen Planungsverfahren in unserem Land. Denn wir haben ja auch große Ausgabereste im Bundeshaushalt. Zweistellige Milliardenbeträge werden regelmäßig nicht abgerufen,"

str2 = "  zweistellige Milliardenbeträge werden regelmäßig nicht abgerufen, die für Investitionen vorgesehen sind. Und das spricht eben auch nochmal dafür,"

str3 = " Weil zweistellige Milliardenbeträge werden regelmäßig nicht abgerufen, die für Investitionen vorgesehen sind. Und das spricht eben auch nochmal dafür,"
```

```python
import Levenshtein

s1 = "zweistellige Milliarden"
s2 = "test zweistellige Milliarden werden abgerufen"
s1 = str1
s2 = str3
```

```python
def _concatenate_strings(s1, s2):
    if not (s1 and s2):
        return s1 + s2
    results = []
    for i in range(1, min(len(s1), len(s2), 250), 1):
        sub1 = s1[-i:]
        sub2 = s2[:i]
        edits = Levenshtein.editops(sub1, sub2)
        case = {"str_len": i, "edits": len(edits), "similarity": i - len(edits)}
        results.append(case)

    best_result = max(results, key=lambda x: x["similarity"])
    return s1 + s2[best_result["str_len"] :]
```
